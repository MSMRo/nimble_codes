## Test 1

Respuesta en base a la intención del query

```python
from langchain_core.runnables import RunnableLambda
```

```python
def nimble_router_node(inputs: dict) -> str:
    """Nodo que evalúa la consulta usando nimble y el dataset de temperaturas."""
    prompt = inputs.get("input", "")

    # Combinamos la tabla de datos y la pregunta del usuario dentro del 'state'
    state_data = {
        "registro_temperaturas": tabla_temperaturas,
        "consulta_usuario": prompt,
    }

    questions = {
        "intent": Choice(
            instructions=(
                "Basándote en la tabla de temperaturas y la consulta del"
                " usuario, clasifica el tipo de pregunta."
            ),
            criteria={
                "temp_max": (
                    "Preguntas sobre valores máximos, picos de calor o si se"
                    " superó cierta temperatura."
                ),
                "rango_horario": (
                    "Preguntas sobre horas específicas o momentos del día"
                    " (mañana, tarde, etc.)."
                ),
                "otro": "Preguntas generales que no se relacionan con los datos de temperatura.",
            },
        )
    }

    with TypeSafeClient() as client:
        res = client.system_one(state=state_data, questions=questions)

    return res.choices["intent"].choice
```

```python

# 2. Crear el runnable de LangChain
router_runnable = RunnableLambda(nimble_router_node)

# 3. Probar la decisión
decision = router_runnable.invoke(
    {"input": "¿los 24 grados en algún momento del día?"}
)
print("Ruta seleccionada por LangChain:", decision)  # Devuelve 'temp_max'
```

## Test 2

### Definición del esquema con Pydantic (schema.py)
Definimos la estructura estricta que obligará a Ollama a retornar únicamente los datos tipados de la consulta (métrica, función de agregación y rango horario en formato ISO).

Python

```python
%%writefile schema.py

from datetime import datetime
from typing import Literal, Optional
from pydantic import BaseModel, Field


class QuerySensorsIntent(BaseModel):
    """Estructura para extraer parámetros de consulta sobre series temporales de sensores."""

    metric: Literal["temperature", "humidity", "pressure"] = Field(
        default="temperature",
        description="Métrica a consultar (temperature, humidity, pressure).",
    )
    aggregation: Literal["MAX", "MIN", "AVG", "NONE"] = Field(
        default="NONE",
        description="Operación matemática solicitada: MAX, MIN, AVG o NONE si se pide la serie completa.",
    )
    start_time: datetime = Field(
        description="Marca de tiempo de inicio en formato ISO 8601 (YYYY-MM-DDTHH:MM:SS)."
    )
    end_time: datetime = Field(
        description="Marca de tiempo de fin en formato ISO 8601 (YYYY-MM-DDTHH:MM:SS)."
    )
    device_id: Optional[str] = Field(
        default=None,
        description="Identificador del sensor o dispositivo si se especifica.",
    )
```

### Pipeline completo con LangChain (main.py)
Este script configura el modelo en Ollama con salida estructurada (with_structured_output), ejecuta la consulta SQL parametrizada mediante SQLAlchemy y genera la respuesta final.

```python
import os
from datetime import datetime
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from pydantic import BaseModel, Field
from sqlalchemy import create_engine, text
from schema import QuerySensorsIntent
```

```python

# ==========================================
# 1. Configuración de Base de Datos
# ==========================================
# Para pruebas locales puedes usar SQLite; si usas PostgreSQL reemplaza por:
# "postgresql+psycopg2://usuario:password@localhost:5432/tu_db"
DATABASE_URL = "sqlite:///sensores.db"
engine = create_engine(DATABASE_URL)

def setup_test_db():
    """Crea una tabla de prueba y datos dummy si no existe."""
    with engine.connect() as conn:
        conn.execute(
            text(
                """
            CREATE TABLE IF NOT EXISTS lecturas_sensores (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                device_id TEXT,
                temperatura REAL,
                humedad REAL,
                fecha_hora DATETIME
            );
        """
            )
        )
        # Insertar algunos registros de prueba para hoy
        conn.execute(
            text(
                """
            INSERT INTO lecturas_sensores (device_id, temperatura, humedad, fecha_hora)
            VALUES 
                ('esp32_01', 22.4, 60.0, '2026-10-15 14:05:00'),
                ('esp32_01', 28.7, 55.0, '2026-10-15 15:30:00'),
                ('esp32_01', 24.1, 58.0, '2026-10-15 17:15:00');
        """
            )
        )
        conn.commit()
```

```python
# ==========================================
# 2. Inicialización del Modelo Ollama
# ==========================================
# Asegúrate de que Ollama esté corriendo (ollama serve)
llm = ChatOllama(
    model="nimble",  # o el modelo que estés corriendo en tu instancia
    temperature=0.0,  # 0 para máxima consistencia y fidelidad al esquema
)
```

```python
# Forzamos la salida estructurada según nuestro modelo Pydantic
structured_llm = llm.with_structured_output(QuerySensorsIntent)

# Plantilla con contexto temporal (clave para resolver términos como 'hoy', 'ayer', 'hace una hora')
system_prompt = """Eres un extractor semántico para un sistema de IoT y sensores.
Tu único trabajo es mapear la pregunta del usuario al esquema estructurado de consulta.
La fecha y hora actual de referencia es: {current_time}.
No inventes datos que el usuario no mencione explícitamente."""

prompt_template = ChatPromptTemplate.from_messages(
    [("system", system_prompt), ("human", "{user_query}")]
)

# Cadena de extracción
extraction_chain = prompt_template | structured_llm
```

```python
# ==========================================
# 3. Capa de Ejecución Determinista (SQL)
# ==========================================
def ejecutar_en_base_de_datos(params: QuerySensorsIntent):
    """Traduce el objeto estructurado a una sentencia SQL segura y parametrizada."""
    print(f"\n[INFO] Parámetros extraídos por el modelo:\n{params.model_dump_json(indent=2)}")

    column_map = {
        "temperature": "temperatura",
        "humidity": "humedad",
    }
    columna = column_map.get(params.metric, "temperatura")

    with engine.connect() as conn:
        if params.aggregation in ["MAX", "MIN", "AVG"]:
            sql = f"""
                SELECT {params.aggregation}({columna}) AS resultado
                FROM lecturas_sensores
                WHERE fecha_hora BETWEEN :start_time AND :end_time
            """
            result = conn.execute(
                text(sql),
                {
                    "start_time": params.start_time.strftime("%Y-%m-%d %H:%M:%S"),
                    "end_time": params.end_time.strftime("%Y-%m-%d %H:%M:%S"),
                },
            )
            valor = result.scalar()
            return f"El valor {params.aggregation} de {params.metric} entre {params.start_time} y {params.end_time} fue: {valor}"

        else:
            # Caso donde pide todas las lecturas del rango horario
            sql = f"""
                SELECT fecha_hora, {columna}
                FROM lecturas_sensores
                WHERE fecha_hora BETWEEN :start_time AND :end_time
                ORDER BY fecha_hora ASC
            """
            result = conn.execute(
                text(sql),
                {
                    "start_time": params.start_time.strftime("%Y-%m-%d %H:%M:%S"),
                    "end_time": params.end_time.strftime("%Y-%m-%d %H:%M:%S"),
                },
            )
            filas = result.fetchall()
            return [dict(row._mapping) for row in filas]

```

```python
# ==========================================
# 4. Flujo Principal de Ejecución
# ==========================================
if __name__ == "__main__":
    setup_test_db()

    # Consulta del usuario
    pregunta = "¿Cuál fue la temperatura máxima hoy entre las 14:00 y las 18:00?"

    # 1. Extraer entidades y filtros con Ollama + LangChain
    hora_actual = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    parametros_query: QuerySensorsIntent = extraction_chain.invoke(
        {"user_query": pregunta, "current_time": hora_actual}
    )

    # 2. Consultar la base de datos real con los parámetros limpios
    resultado_final = ejecutar_en_base_de_datos(parametros_query)

    # 3. Mostrar resultado computado real
    print(f"\n[RESULTADO BD]:\n{resultado_final}")
```
